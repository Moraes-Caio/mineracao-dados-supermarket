# Carregando .ARFF

```python
from scipy.io.arff import loadarff
import numpy as np
import pandas as pd
```

```python
data, meta = loadarff('/content/supermarket.arff')
data_np = np.array(data)

df = pd.DataFrame(data_np, dtype = str)
df.to_csv('supermarket.csv', index=False)
df
```

```python
df_nan = pd.read_csv('/content/supermarket.csv', na_values=['?'])
df_nan
```

```python
info_counts = df_nan.info(verbose=True, show_counts=True)
```

```python
df_nan.department7.isna().all()
```

```python
colunas_removerf = []
i = 0

while (i < len(df_nan.columns)):

    if (df_nan.iloc[:, i].isna().all() == np.True_):
        colunas_removerf.append(df_nan.columns[i])


    i = i + 1

print(colunas_removerf)

df_filtrado = df_nan.drop(colunas_removerf, axis=1)
df_filtrado
```

```python
colunas = df_filtrado.columns
colunas_remover1 = []
i = 0

while i < len(colunas):
    coluna_atual = colunas[i]

    if coluna_atual.startswith('depart'):
        colunas_remover1.append(coluna_atual)


    i = i + 1

print(colunas_remover1)

df_t= df_filtrado.drop(colunas_remover1, axis=1)
df_t
```

```python
df_t.info(verbose=True, show_counts=True)
```

```python
df_t.to_csv('supermarket_filtrado.csv', index=False)
```
